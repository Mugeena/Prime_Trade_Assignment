```python
import pandas as pd
trader_data = pd.read_csv(r"C:\Users\mukin\Downloads\historical_data.csv")
fear_greed = pd.read_csv(r"C:\Users\mukin\Downloads\fear_greed_index.csv")
print("Trader Data")
print(trader_data.head())
print("\nFear & Greed Data")
print(fear_greed.head())
```

```python
print(trader_data.columns)
print(fear_greed.columns)
```

```python
print("Trader Data Shape:", trader_data.shape)
print("Fear & Greed Shape:", fear_greed.shape)
```

```python
print(trader_data.isnull().sum())
print(fear_greed.isnull().sum())
```

```python
# Convert trader dates
trader_data['Timestamp IST'] = pd.to_datetime(
    trader_data['Timestamp IST'],
    format='%d-%m-%Y %H:%M'
)

trader_data['date'] = trader_data['Timestamp IST'].dt.date

# Convert sentiment dates
fear_greed['date'] = pd.to_datetime(
    fear_greed['date']
).dt.date

# Merge datasets
merged_data = pd.merge(
    trader_data,
    fear_greed[['date', 'classification', 'value']],
    on='date',
    how='left'
)
```

```python
print("Merged Shape:", merged_data.shape)

print(merged_data[['date', 'classification']].head())

print(merged_data['classification'].value_counts())
```

```python
print(merged_data.shape)
print(merged_data['classification'].value_counts())
```

```python
pnl_analysis = merged_data.groupby('classification')['Closed PnL'].agg([
    'count',
    'sum',
    'mean',
    'max',
    'min'
])

print(pnl_analysis)
```

```python
win_rate = merged_data.groupby('classification').apply(
    lambda x: (x['Closed PnL'] > 0).mean() * 100
)

print(win_rate)
```

```python
import matplotlib.pyplot as plt

avg_pnl = merged_data.groupby('classification')['Closed PnL'].mean()

avg_pnl.plot(kind='bar')

plt.title('Average Profit/Loss by Sentiment')
plt.ylabel('Average Closed PnL')
plt.show()
```

```python
print(pnl_analysis)

print(win_rate)
```

```python
trade_volume = merged_data.groupby('classification').agg({
    'Size USD':['count','sum','mean']
})

print(trade_volume)
```

```python
top_traders = merged_data.groupby('Account')['Closed PnL'].sum()

top_traders = top_traders.sort_values(
    ascending=False
).head(10)

print(top_traders)
```

```python
import matplotlib.pyplot as plt

top_traders.plot(kind='bar')

plt.title("Top 10 Traders by Total Profit")
plt.ylabel("Total Closed PnL")
plt.xlabel("Trader Account")
plt.show()
```

```python
win_rate.plot(kind='bar')

plt.title("Win Rate by Sentiment")
plt.ylabel("Win Rate (%)")
plt.xlabel("Sentiment")
plt.show()
```
